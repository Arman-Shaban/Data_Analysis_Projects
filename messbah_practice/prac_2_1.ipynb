```python
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt

sns.set_theme(style='whitegrid')
%matplotlib inline
```

```python
df = sns.load_dataset("titanic")
df.head()
```

```python
print('Dataset dimensions, rows and columns:')
print(df.shape)
```

```python
df.info()
```

```python
print('Number of missing values:')
print(df.isnull().sum())
```

```python
plt.figure(figsize=(10,4))
sns.heatmap(df.isnull(), cbar=False, yticklabels=True, cmap='viridis')
```

```python
df['age'] = df['age'].fillna(df['age'].median())

df_clean = df.drop(columns=['deck'])
print('missing values after cleaning:')
print(df_clean.isnull().sum())
```

```python
print('Survival rate by sex:')
print(df_clean.groupby('sex')['survived'].mean())
```

```python
print('Survival rate by ticket class:')
print(df_clean.groupby('pclass')['survived'].mean())
```

```python
print('Combined survival rate:')
print(df_clean.pivot_table(
    values='survived',
    index='sex',
    columns='pclass'
))
```

```python
plt.figure(figsize=(7,5))
sns.barplot(
    data=df_clean,
    x='pclass',
    y='survived',
    hue='sex',
    palette='viridis'
)
plt.title("survival rate by ticket class and sex:")
plt.ylabel('probability of survival')
plt.xlabel('ticket class')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(
    data=df_clean,
    x='age',
    hue ='survived',
    bins = 30,
    palette='coolwarm'
)
plt.title('age distribution by survival status')
plt.ylabel('age')
plt.xlabel('number of passengers')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.barplot(
    data=df_clean,
    x='pclass',
    y='survived',
    hue='sex',
    ax=axes[0],
    palette='viridis'
)

axes[0].set_title('Survival Rate by Ticket Class and Sex')
axes[0].set_ylabel('Probability of Survival')
axes[0].set_xlabel('Ticket Class')

# Second plot: Age distribution by survival status
sns.histplot(
    data=df_clean,
    x='age',
    hue='survived',
    kde=True,
    bins=30,
    ax=axes[1],
    palette='coolwarm'
)

axes[1].set_title('Age Distribution by Survival Status')
axes[1].set_xlabel('Age')
axes[1].set_ylabel('Number of Passengers')

plt.tight_layout()
plt.show()
```
